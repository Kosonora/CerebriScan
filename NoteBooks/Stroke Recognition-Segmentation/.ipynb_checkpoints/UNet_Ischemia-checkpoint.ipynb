```python
#分割数据集
from sklearn.model_selection import train_test_split
TrainPath,SplitPath,TrainLabel,SplitLabel=train_test_split(IschemiaImagePath.loc[:,'PNG'],IschemiaImagePath.loc[:,'MASK'],train_size=0.7,
                                                           random_state=2025)
ValidPath,TestPath,ValidLabel,TestLabel=train_test_split(SplitPath,SplitLabel,train_size=0.5,random_state=2025)
print(f'训练集地址:{len(TrainPath)},验证集地址:{len(ValidPath)},测试集地址:{len(TestPath)}')
print(f'训练集标签:{len(TrainLabel)},验证集标签:{len(ValidLabel)},测试集标签:{len(TestLabel)}')
```

```python
#读取.npy文件
import numpy as np
def read_npy_to_tensor(npy_file_path):
    #读取.npy文件
    if isinstance(npy_file_path, tf.Tensor):
        npy_file_path = npy_file_path.numpy().decode('utf-8')
    npy_data=np.load(npy_file_path)
    tf_tensor = tf.convert_to_tensor(npy_data, dtype=tf.float32)
    return tf_tensor
```

```python
#编写数据集读取函数
import tensorflow as tf
def ReadImageData(ct_path,mask_path):
        CT_io=tf.io.read_file(ct_path)    #读取文件
        CT_data=tf.image.decode_png(CT_io,channels=3)    #解码图像数据
        CT_size=tf.image.resize(CT_data,(512,512))
        CT_normal=tf.cast(CT_size, tf.float32)/255.0
        Mask_tensor=tf.py_function(func=read_npy_to_tensor,inp=[mask_path],Tout=tf.float32)
        Mask_tensor.set_shape((512, 512, 1)) 
        return (CT_normal,Mask_tensor)
```

```python
#使用文件路径方法批量读取数据
#训练集
FilePathDataset_Train=tf.data.Dataset.from_tensor_slices((TrainPath,TrainLabel)).repeat(1)    #文件路径数据集
ImageDataset_Train=FilePathDataset_Train.shuffle(buffer_size=1000,seed=2026).map(ReadImageData,num_parallel_calls=tf.data.experimental.AUTOTUNE)    #获取文件路径
ImageDataset_Train=ImageDataset_Train.cache()
ImageDataset_Train=ImageDataset_Train.batch(batch_size=8).prefetch(tf.data.experimental.AUTOTUNE)
#验证集
FilePathDataset_Valid=tf.data.Dataset.from_tensor_slices((ValidPath,ValidLabel)).repeat(1)    #文件路径数据集
ImageDataset_Valid=FilePathDataset_Valid.shuffle(buffer_size=500,seed=2026).map(ReadImageData,num_parallel_calls=tf.data.experimental.AUTOTUNE)    #获取文件路径
ImageDataset_Valid=ImageDataset_Valid.batch(batch_size=8).prefetch(tf.data.experimental.AUTOTUNE)
#测试集
FilePathDataset_Test=tf.data.Dataset.from_tensor_slices((TestPath,TestLabel)).repeat(1)    #文件路径数据集
ImageDataset_Test=FilePathDataset_Test.shuffle(buffer_size=500,seed=2026).map(ReadImageData,num_parallel_calls=tf.data.experimental.AUTOTUNE)    #获取文件路径
ImageDataset_Test=ImageDataset_Test.batch(batch_size=8).prefetch(tf.data.experimental.AUTOTUNE)
```

构建U-Net网络

```python
#下采样模块
from tensorflow import keras
class EncoderBlock(keras.layers.Layer):
    def __init__(self,filters,rate,activation,pooling=True, **kwargs):
        super(EncoderBlock, self).__init__(**kwargs)
        self.filters=filters   # 卷积核数量（特征图通道数）
        self.rate=rate   # Dropout丢弃率（正则化，防止过拟合）
        self.pooling=pooling    # 是否执行最大池化下采样
        self.activation=activation   #激活函数
        #计算模块
        self.Conv2D1=keras.layers.Conv2D(filters=self.filters,kernel_size=3, strides=1, padding='same', activation=None, kernel_initializer='he_normal')
        self.drop=keras.layers.Dropout(rate=self.rate)
        self.BN1=keras.layers.LayerNormalization(axis=-1)
        self.Act1=keras.layers.Activation(self.activation) 
        self.Conv2D2=keras.layers.Conv2D(filters=self.filters, kernel_size=3, strides=1, padding='same', activation=None, kernel_initializer='he_normal')
        self.BN2 =keras.layers.LayerNormalization(axis=-1) 
        self.Act2 = keras.layers.Activation(self.activation) 
        self.pool=keras.layers.MaxPool2D()
    def call(self,Input):   
        output1=self.Conv2D1(Input)
        bn1=self.BN1(output1)
        ac1=self.Act1(bn1)
        dropou1=self.drop(ac1)
        output2=self.Conv2D2(dropou1)
        bn2=self.BN2(output2)       # 卷积后添加批量归一化
        ac2= self.Act2(bn2)
        if self.pooling:    #返回downsample,ac2是编码器的关键设计：
            downsample=self.pool(ac2)   #downsample是下采样后的 “压缩特征”（语义信息更丰富）
            return downsample,ac2
        else:
            return ac2    #ac2是池化前的 “高分辨率特征”，后者会被解码器拼接，弥补下采样的细节丢失。
    def get_config(self):
        base_config=super().get_config()
        return {**base_config,"filters":self.filters,'rate':self.rate,
                'pooling':self.pooling,'activation':self.activation}
```

```python
#上采样模块
class DecoderBlock(keras.layers.Layer):
    def __init__(self,filters,rate,activation,**kwargs):
        super(DecoderBlock,self).__init__(**kwargs)
        self.filters=filters
        self.rate=rate
        self.activation = activation
        self.up=keras.layers.Conv2DTranspose(filters=self.filters,kernel_size=(2, 2),activation=None,strides=2, padding="same")    # 2×2上采样层
        self.BN_up =keras.layers.LayerNormalization(axis=-1)  # 新增：转置卷积后批量归一化
        self.Act_up = keras.layers.Activation(self.activation)
        self.net=EncoderBlock(filters=self.filters,rate=self.rate,activation=self.activation,pooling=False)   #关闭池化：仅用其“双卷积+Dropout”逻辑
    def call(self,inputs):
        X,skip_X=inputs    # 输入X是一个元组：当前解码器输入特征, 编码器对应层的跳跃连接特征
        Out_up=self.up(X)   #上采样
        Out_bn=self.BN_up(Out_up)
        Out_ac=self.Act_up(Out_bn)
        Out_Concat=keras.layers.concatenate([Out_ac,skip_X])   #拼接
        Out=self.net(Out_Concat)    #执行卷积
        return Out
    def get_config(self):
        base_config=super().get_config()
        return {**base_config,"filters":self.filters,'rate':self.rate,'activation':self.activation}
```

```python
#构建UNet模型
def UNet(input_shape,num_classes):
    inputs=keras.layers.Input(shape=input_shape,name='Input')   #输入层
    #编码器:下采样
    pool1,conv1=EncoderBlock(filters=64,rate=0.5,pooling=True,activation='relu',name='EncoderBlock1')(inputs)    #第1层:512x512 → 256x256
    pool2,conv2=EncoderBlock(filters=128,rate=0.5,pooling=True,activation='relu',name='EncoderBlock2')(pool1)      # 第2层：256x256 → 128x128
    pool3,conv3=EncoderBlock(filters=256,rate=0.5,pooling=True,activation='relu',name='EncoderBlock3')(pool2)   # 第3层：128x128 → 64x64
    pool4,conv4=EncoderBlock(filters=512,rate=0.5,pooling=True,activation='relu',name='EncoderBlock4')(pool3)     # 第4层：64x64 → 32x32
    #瓶颈层
    bottleneck=EncoderBlock(filters=1024,rate=0.5,pooling=False,activation='relu',name='EncoderBlock5')(pool4)  # 32x32x1024
    #解码器:上采样
    dec4=DecoderBlock(filters=512,rate=0.5,activation='relu',name='DncoderBlock1')([bottleneck,conv4])  # 第4层：32x32 → 64x64（拼接conv4）
    dec3=DecoderBlock(filters=256,rate=0.5,activation='relu',name='DncoderBlock2')([dec4,conv3])   # 第3层：64x64 → 128x128（拼接conv3）
    dec2=DecoderBlock(filters=128,rate=0.5,activation='relu',name='DncoderBlock3')([dec3,conv2])   # 第2层：128x128 → 256x256（拼接conv2）
    dec1=DecoderBlock(filters=64,rate=0.5,activation='relu',name='DncoderBlock4')([dec2, conv1])   # 第1层：256x256 → 512x512（拼接conv1）
    #输出层   1x1卷积将特征映射为单通道掩码（512x512x1）
    outputs=keras.layers.Conv2D(num_classes, (1, 1), activation='sigmoid',name='Output')(dec1)
    U_Net=keras.models.Model(inputs=inputs, outputs=outputs, name="U-Net")
    return U_Net
```

基于脑部CT区域：正常和缺血统计指标，进行损失函数设计和优化。

```python
#统计掩码区域占比，查看是否存在类别不均衡
import numpy as np
import gc
def calculate_mask_region_ratio(mask):
    # 1. 输入形状验证
    if mask.shape != (512, 512,1):
        raise ValueError(f"mask数组形状必须为(512,512)，当前输入形状为{mask.shape}")
    # 2. 检查是否存在非0/1的值
    unique_vals = np.unique(mask)
    if not np.all(np.isin(unique_vals, [0, 1])):
        import warnings
        warnings.warn(f"mask数组中检测到非0/1的像素值：{unique_vals}，仅统计0和1的比例", UserWarning)
    #计算总像素数、各区域像素数
    mask=mask.reshape(512,512)
    total_pixels = mask.size 
    segment_pixel_count = np.sum(mask == 1)  # 分割区域（1）的像素数
    normal_pixel_count = total_pixels - segment_pixel_count  # 正常区域（0）的像素数
    #计算比例
    normal_ratio = round(normal_pixel_count / (total_pixels + 1e-8), 4)
    segment_ratio = round(segment_pixel_count / (total_pixels + 1e-8), 4)
    #内存管理
    del mask,total_pixels,segment_pixel_count,normal_pixel_count
    gc.collect()
    # 5. 返回统计结果
    return normal_ratio,segment_ratio
```

```python
import tqdm
Normal_rate,Ischemia_Rate=[],[]
for mask_path in tqdm.tqdm(IschemiaImagePath['MASK']):
    mask=np.load(mask_path)
    normal_ratio,segment_ratio=calculate_mask_region_ratio(mask=mask)
    Normal_rate.append(normal_ratio)
    Ischemia_Rate.append(segment_ratio)
```

```python
# 绘制直方图
color_normal = '#1f77b4'  # 蓝色（正常）
color_bleeding = '#d62728'  # 红色（出血）
plt.figure()
plt.hist(Normal_rate, alpha=0.6, label='Normal area', color=color_normal, density=True)
plt.hist(Ischemia_Rate, alpha=0.6, label='Ischemia area', color=color_bleeding, density=True)
plt.title('histogram', fontsize=14, fontweight='bold')
plt.xlabel('Mean value (after normalisation)', fontsize=12)
plt.ylabel('density', fontsize=12)
plt.legend()
plt.grid(axis='y', alpha=0.3)
```

```python
import tensorflow as tf
from tensorflow.keras import losses
class BCE_DiceLoss(losses.Loss):
    def __init__(self, alpha=0.5, power=3, smooth=1e-8, from_logits=False,reduction=keras.losses.Reduction.AUTO,name="bce_dice_loss"):
        super().__init__(reduction=reduction,name=name)
        self.alpha = alpha
        self.power = power
        self.smooth = smooth
        self.from_logits = from_logits
        self.bce = losses.BinaryCrossentropy(from_logits=from_logits, reduction=losses.Reduction.AUTO)
    def call(self, y_true, y_pred):
        # 1. 计算BCE损失
        bce_loss = self.bce(y_true, y_pred)
        bce_loss = tf.reduce_mean(bce_loss)
        # 2. 计算改进Dice损失
        if self.from_logits:
            y_pred = tf.nn.sigmoid(y_pred)
        # 展平为一维（适配任意尺寸图像）
        y_true_flat = tf.keras.backend.flatten(y_true)
        y_pred_flat = tf.keras.backend.flatten(y_pred)
        # 改进Dice：预测值取幂次，强化前景/困难样本
        intersection = tf.reduce_sum(tf.pow(y_pred_flat, self.power) * y_true_flat)
        union = tf.reduce_sum(tf.pow(y_pred_flat, self.power)) + tf.reduce_sum(y_true_flat)
        dice_loss = 1.0 - (2.0 * intersection + self.smooth) / (union + self.smooth)
        # 3. 组合损失
        total_loss = self.alpha * bce_loss + (1 - self.alpha) * dice_loss
        return total_loss

    def get_config(self):
        config = super().get_config()
        config.update({
            "alpha": self.alpha,
            "power": self.power,
            "smooth": self.smooth,
            "from_logits": self.from_logits
        })
        return config
```

自定义评估指标

```python
import tensorflow as tf
from tensorflow.keras import metrics
class BinaryIoUMetrics(metrics.Metric):
    def __init__(self,threshold=0.5,smooth=1e-8,name="binary_iou_metrics",**kwargs):
        super().__init__(name=name, **kwargs)
        self.threshold = threshold
        self.smooth = smooth
        # 初始化累计变量：分别累计正/负类的交集、并集
        # 正类（标签1）
        self.true_positives = self.add_weight(name="tp", initializer="zeros")  # 正类交集
        self.false_negatives = self.add_weight(name="fn", initializer="zeros")
        self.false_positives = self.add_weight(name="fp", initializer="zeros")
        # 负类（标签0）
        self.true_negatives = self.add_weight(name="tn", initializer="zeros")  # 负类交集
        self.false_positives_neg = self.add_weight(name="fp_neg", initializer="zeros")
        self.false_negatives_neg = self.add_weight(name="fn_neg", initializer="zeros")
        #累计每个batch的交并集数据
    def update_state(self, y_true, y_pred, sample_weight=None):
        y_pred_binary = tf.cast(y_pred >= self.threshold, tf.float32)
        y_true = tf.cast(y_true, tf.float32)
        # 展平张量
        y_true_flat = tf.keras.backend.flatten(y_true)
        y_pred_flat = tf.keras.backend.flatten(y_pred_binary)
        # 2. 计算正类（标签1）的交并集
        # 正类交集（TP）：真实1且预测1
        tp = tf.reduce_sum(y_true_flat * y_pred_flat)
        # 正类假负（FN）：真实1但预测0
        fn = tf.reduce_sum(y_true_flat * (1 - y_pred_flat))
        # 正类假正（FP）：真实0但预测1
        fp = tf.reduce_sum((1 - y_true_flat) * y_pred_flat)
        # 3. 计算负类（标签0）的交并集
        # 负类交集（TN）：真实0且预测0
        tn = tf.reduce_sum((1 - y_true_flat) * (1 - y_pred_flat))
        # 负类假负（FN_neg）：真实0但预测1（即正类FP）
        fn_neg = fp
        # 负类假正（FP_neg）：真实1但预测0（即正类FN）
        fp_neg = fn
        # 4. 累计到全局变量（支持sample_weight，可选）
        if sample_weight is not None:
            sample_weight = tf.cast(sample_weight, tf.float32)
            sample_weight = tf.keras.backend.flatten(sample_weight)
            tp = tf.reduce_sum(sample_weight * tp)
            fn = tf.reduce_sum(sample_weight * fn)
            fp = tf.reduce_sum(sample_weight * fp)
            tn = tf.reduce_sum(sample_weight * tn)
            fn_neg = tf.reduce_sum(sample_weight * fn_neg)
            fp_neg = tf.reduce_sum(sample_weight * fp_neg)
        # 更新累计值
        self.true_positives.assign_add(tp)
        self.false_negatives.assign_add(fn)
        self.false_positives.assign_add(fp)
        self.true_negatives.assign_add(tn)
        self.false_positives_neg.assign_add(fp_neg)
        self.false_negatives_neg.assign_add(fn_neg)
        #计算最终指标：正类IoU、负类IoU、平均IoU
    def result(self):
        #正类IoU = TP / (TP + FN + FP)
        iou_pos = (self.true_positives + self.smooth) / (
            self.true_positives + self.false_negatives + self.false_positives + self.smooth
        )
        #负类IoU = TN / (TN + FN_neg + FP_neg)
        iou_neg = (self.true_negatives + self.smooth) / (
            self.true_negatives + self.false_negatives_neg + self.false_positives_neg + self.smooth
        )
        #平均IoU（算术平均）
        iou_avg = (iou_pos + iou_neg) / 2.0
        
        #返回字典，方便训练时查看各指标
        return {
            "iou_positive": iou_pos,   # 正类IoU（分割区域）
            "iou_negative": iou_neg,   # 负类IoU（正常区域）
            "iou_average": iou_avg     # 平均IoU
        }
    #重置累计变量（每个epoch/验证周期后调用）
    def reset_states(self):
        self.true_positives.assign(0.0)
        self.false_negatives.assign(0.0)
        self.false_positives.assign(0.0)
        self.true_negatives.assign(0.0)
        self.false_positives_neg.assign(0.0)
        self.false_negatives_neg.assign(0.0)
        #保存配置
    def get_config(self):
        """"""
        config = super().get_config()
        config.update({
            "threshold": self.threshold,
            "smooth": self.smooth
        })
        return config
```

```python
#绘制训练曲线图
import matplotlib.pyplot as plt
def DisplayCurve(history,epoch):
    fig=plt.figure(figsize=(14,18)) 
    ax1=fig.add_subplot(3,1,1) #绘制误差曲线 
    ax1.set_title('loss Curve') 
    ax1.grid(color='black',linestyle='-.',alpha=0.3) 
    ax1.plot(np.arange(1,epoch+1),history['loss'],
             color='blue',linestyle='-',label='Train loss=%f'%history.loc[:,'loss'].min()) 
    ax1.plot(np.arange(1,epoch+1),history['val_loss'],
             color='orange',linestyle='-',label='Valid loss=%f'%history.loc[:,'val_loss'].min()) 
    ax1.set_xlabel('Train epochs') 
    ax1.set_ylabel('loss') 
    ax1.legend()
    ax2=fig.add_subplot(3,1,2) #绘制精度曲线 
    ax2.set_title('Accuracy Curve') 
    ax2.grid(color='black',linestyle='-.',alpha=0.3) 
    ax2.plot(np.arange(1,epoch+1),history['accuracy'],color='blue',linestyle='-',
            label='Train Accuracy=%f'%history['accuracy'].max()) 
    ax2.plot(np.arange(1,epoch+1),history['val_accuracy'],color='orange',linestyle='-',
            label='Valid Accuracy=%f'%history['val_accuracy'].max()) 
    ax2.set_xlabel('Train epochs') 
    ax2.set_ylabel('Accuracy') 
    ax2.legend()
    ax3=fig.add_subplot(3,1,3) #绘制IoU曲线 
    ax3.set_title('IoU Curve') 
    ax3.grid(color='black',linestyle='-.',alpha=0.3) 
    ax3.plot(np.arange(1,epoch+1),history['MeanIoU'],
             color='blue',linestyle='-',label='Train IoU=%f'%history['MeanIoU'].max()) 
    ax3.plot(np.arange(1,epoch+1),history['val_MeanIoU'],
             color='orange',linestyle='-',label='Valid IoU=%f'%history['val_MeanIoU'].max()) 
    ax3.set_xlabel('Train epochs') 
    ax3.set_ylabel('IoU') 
    ax3.legend()
    plt.show() 
```

```python
#手动清理TensorFlow显存和内存的辅助函数
import gc
def clear_gpu_memory():
    #清空Keras会话（释放模型相关的显存占用）
    K.clear_session()
    #强制Python垃圾回收（释放内存中的临时变量）
    gc.collect()
    #重置GPU显存统计（清理显存缓存，无副作用）
    try:
        tf.config.experimental.reset_memory_stats('GPU:0')
    except Exception as e:
        print(f"重置显存统计失败：{e}")
```

```python
#批量计算预测标签
import tqdm
def PredictComputeGPU(Model, Dataset):
    # 存储所有批次结果的列表（使用tf.Tensor避免Python对象内存膨胀）
    true_labels=[]
    pred_proba=[]
    # 遍历数据集批次
    for data_batch,label_batch in tqdm.tqdm(Dataset):
        with tf.device('/GPU:0'):  # 显式指定GPU加速
            predict_proba=Model(data_batch, training=False) 
        # 转换为numpy并存储（批量操作而非单样本）
        true_labels.append(label_batch.numpy())
        pred_proba.append(predict_proba.numpy())
    Label=np.concatenate(true_labels, axis=0)
    PredictLabel=np.concatenate(pred_proba, axis=0)
    #推理完成后释放内存以及显存
    del predict_proba,label_batch,true_labels,pred_proba
    clear_gpu_memory()
    return Label, PredictLabel
```

```python
import numpy as np
import tensorflow as tf
from keras import backend as K
from tensorflow.keras.metrics import AUC, MeanIoU
def calculate_unet_binary_metrics_keras(y_true: np.ndarray,y_pred_prob: np.ndarray,
                                        threshold: float = 0.5,num_classes: int = 2) -> dict:
    #输入预处理（完全匹配MeanIoU的输入要求
    y_true=np.squeeze(y_true)
    y_pred_prob=np.squeeze(y_pred_prob)
    if y_true.shape != y_pred_prob.shape:
        raise ValueError(f"真实标签形状 {y_true.shape} 与预测概率形状 {y_pred_prob.shape} 不一致！")
    # 转换为整数型
    y_true_int = y_true.astype(np.int32)
    y_pred_prob_k = tf.convert_to_tensor(y_pred_prob, dtype=tf.float32)
    # 阈值化得到预测标签
    y_pred_int = tf.cast(tf.greater(y_pred_prob_k, threshold), dtype=tf.int32).numpy()
    # 展平为一维
    y_true_flat = y_true_int.reshape(-1)
    y_pred_flat = y_pred_int.reshape(-1)
    #直接调用MeanIoU计算
    mean_iou_metric=MeanIoU(num_classes=num_classes)
    mean_iou_metric.update_state(y_true_flat, y_pred_flat)
    mean_iou_value=mean_iou_metric.result().numpy()
    # MeanIoU的计算逻辑
    confusion_mtx=tf.math.confusion_matrix(y_true_flat, y_pred_flat,
                                           num_classes=num_classes,dtype=tf.float32).numpy()
    #按类别计算IoU
    iou_per_class = []
    for cls in range(num_classes):
        # 该类别的TP：混淆矩阵对角线元素
        tp = confusion_mtx[cls, cls]
        # 该类别的FP：列和 - TP（所有预测为该类但真实不是的）
        fp = np.sum(confusion_mtx[:, cls]) - tp
        # 该类别的FN：行和 - TP（所有真实为该类但预测不是的）
        fn = np.sum(confusion_mtx[cls, :]) - tp
        # 该类别的IoU（MeanIoU的计算逻辑：tp / (tp + fp + fn)，无eps但自动处理0）
        cls_iou = tp / (tp + fp + fn) if (tp + fp + fn) > 0 else 0.0
        iou_per_class.append(cls_iou)
    #计算所有类别IoU的平均值（MeanIoU的最终结果）
    manual_mean_iou = np.mean(iou_per_class)
    tn, fp, fn, tp = confusion_mtx.ravel() if num_classes == 2 else (0,0,0,0)
    eps = 1e-7
    pixel_acc = (tp + tn) / (tp + tn + fp + fn + eps)
    dice = 2 * tp / (2 * tp + fp + fn + eps)
    precision = tp / (tp + fp + eps)
    recall = tp / (tp + fn + eps)
    specificity = tn / (tn + fp + eps)
    f1 = 2 * (precision * recall) / (precision + recall + eps)
    try:       # AUC计算
        auc_metric = AUC(num_thresholds=1000)
        auc_metric.update_state(y_true_flat, y_pred_prob.reshape(-1))
        auc = auc_metric.result().numpy()
    except ValueError:
        auc = np.nan
    #理结果 
    metrics_result = {"阈值": round(threshold, 2),"mean_iou": round(float(mean_iou_value), 4),
        "mean_iou(多类别)": round(float(manual_mean_iou), 4),
        "iou_per_class": [round(float(i), 4) for i in iou_per_class],  # 类别0/1的IoU
        "dice_coefficient": round(float(dice), 4),
        "pixel_accuracy": round(float(pixel_acc), 4),
        "precision": round(float(precision), 4),"recall_sensitivity": round(float(recall), 4),
        "f1_score": round(float(f1), 4),"specificity": round(float(specificity), 4),
        "auc": round(float(auc), 4) if not np.isnan(auc) else "N/A",
        "confusion_matrix": {
            "tp": int(tp), "tn": int(tn), "fp": int(fp), "fn": int(fn),
            "full_confusion_matrix": confusion_mtx.tolist()  # 全类别混淆矩阵
        }
    }
    # 清理资源
    del mean_iou_metric, confusion_mtx
    K.clear_session()
    return metrics_result
```

```python
#训练集
train_label,train_proba=PredictComputeGPU(Model=IschemiaUNet, Dataset=ImageDataset_Train)
calculate_unet_binary_metrics_keras(y_true=train_label,y_pred_prob=train_proba,
                                    threshold=0.5)
```

```python
#验证集
valid_label,valid_proba=PredictComputeGPU(Model=IschemiaUNet, Dataset=ImageDataset_Valid)
calculate_unet_binary_metrics_keras(y_true=valid_label,y_pred_prob=valid_proba,
                                    threshold=0.5)
```

```python
#对比真实标签
import cv2
def ShowSagemen(ct_path,mask_path,model,threshold=0.5):
    Image_raw=cv2.imread(ct_path)
    Image_resize=cv2.resize(Image_raw,dsize=(512,512))
    Image_Norm=Image_resize/255
    image_d=np.expand_dims(Image_Norm, axis=0)
    #读取真实分割
    Mask=np.load(mask_path)
    predit_proba=model.predict(image_d)   #获取模型预测结果
    predict_mask=(predit_proba>threshold).astype(np.int32)    #标签二值化
    #显示预测结果
    fig=plt.figure(figsize=(18,5),dpi=300)   #可视化模型预测结果
    ax1=fig.add_subplot(1,3,1)   # 原始图像
    ax1.set_title('Original segmentation area')
    ax1.imshow(Image_Norm)
    ax1.imshow(Mask,alpha=0.4)
    cmap=plt.cm.get_cmap('jet')
    ax2=fig.add_subplot(1,3,2)   #预测分割区域
    ax2.set_title('Predicted segmentation area')
    ax2.imshow(Image_Norm)
    proba=ax2.imshow(np.squeeze(predit_proba),alpha=0.5,cmap='jet')
    cbar=fig.colorbar(proba,ax=ax2)    # 添加颜色条
    cbar.set_label('MASK Prediction Probability')
    ax3=fig.add_subplot(1,3,3)   #标签化分割区域
    ax3.set_title('Predictive segmentation area (labelling)')
    ax3.imshow(Image_Norm)
    ax3.imshow(np.squeeze(predict_mask),alpha=0.5)
    plt.tight_layout()
    plt.show()
```

```python
#使用GradCAM可视化模型预测结果
import cv2
import matplotlib.pyplot as plt
from tf_explain.core.grad_cam import GradCAM
def ShowAttention(image_path,model):
    image=cv2.imread(image_path)
    image=cv2.resize(image,dsize=(512,512))
    image_n=image/255
    image_d=np.expand_dims(image_n, axis=0)  
    mask=model.predict(image_d)
    pred_mask = (mask>0.5).astype('float')
    exp = GradCAM()   #实例化GradCAM解释器
    #生成GradCAM热力图
    cam = exp.explain(validation_data=(image_d,mask),class_index=1,layer_name='DncoderBlock4',
                      model=model)
    fig=plt.figure(figsize=(18,5),dpi=300)   #可视化模型预测结果
    ax1=fig.add_subplot(1,3,1)   # 原始图像
    ax1.set_title('CT Image')
    ax1.imshow(image)
    cmap=plt.cm.get_cmap('jet')
    ax2=fig.add_subplot(1,3,2)
    ax2.set_title('Predicted Mask')
    ax2.imshow(image)
    mask=ax2.imshow(np.squeeze(mask),alpha=0.5,cmap='jet')
    mask_cbar=plt.colorbar(mask,ax=ax2)
    mask_cbar.set_label('MASK Prediction Probability')
    #显示GradCam
    ax3=fig.add_subplot(1,3,3)
    ax3.set_title('GradCAM')
    cam=ax3.imshow(cam, cmap='jet',alpha=1.0)
    cbar=fig.colorbar(cam,ax=ax3)    # 添加颜色条
    cbar.set_label('GradCAM')
    plt.tight_layout()
    plt.show()
```
